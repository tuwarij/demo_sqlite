```python
#setup
import chromadb
# client จะคล้ายๆ cursor
client = chromadb.Client()
collection = client.get_or_create_collection(name="demo_collection")
```

```python
# เพิมข้อมูลเข้าไปใน collection
docs = [
    ("1", "The quick brown fox jumps over the lazy dog"),
    ("2", "The lazy dog is sleeping"),
    ("3", "The mechanical keyboard is loud"),
    ("4", "Pc gaming is fun"),
]

ids = [doc[0] for doc in docs]
documents = [doc[1] for doc in docs]

# กันรันซ้ำเฉยๆ
try:
    collection.delete(ids=ids)  
except Exception:
    pass

# มันจะแปลงเป็น embedding ให้เองอัตโนมัติ
collection.add(ids=ids, documents=documents)
```

```python
# ลอง query ดู
query = "pc"
result = collection.query(
    query_texts=[query],
    n_results=2,
    include=["distances", "documents"],
)

print(f"Query: {query!r}")

docs_out = result.get("documents")[0]
dists_out = result.get("distances")[0]
for doc, dist in zip(docs_out, dists_out):
    print(f"- distance={dist:.6f} content={doc}")
```
